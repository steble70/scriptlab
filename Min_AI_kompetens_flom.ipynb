## Bakgrund

Jag byggde ursprungligen `flom.py` som ett litet skript för att generera slumpmässiga politiska floskler. Den första versionen (0.1) läste in fördefinierade ord ur en JSON-fil och slumpade fram en mening enligt mallen «Vi {subjekt} {predikat} {objekt}.» Nedan syns den ursprungliga koden tillsammans med en körning i terminalen.

````python
# flom.py 0.1
# © Stefan Blecko 2025
import json
import random
import os
import argparse


def flom_load(fpath='flom_data.json'):
    os.chdir(os.environ['USERPROFILE'])
    with open(fpath, encoding='utf8') as f:
        unformatted_json = json.load(f)
    jason_data = unformatted_json['floskel_data']
    subjekt = random.choice(jason_data['subjekt'])
    predikat = random.choice(jason_data['predikat'])
    objekt = random.choice(jason_data['objekt'])
    sentence = f"Vi {subjekt} {predikat} {objekt}."
    return sentence


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--antal_floskler", default=1,
                         type=int, help="Antal floskler som ska genereras")
    args = parser.parse_args()

    for _ in range(args.antal_floskler):
        print(flom_load())


if __name__ == "__main__":
    main()
````

**Exempel på körning** (`python flom.py --antal_floskler 5`):

```text
Vi Centerpartiet vill se tillväxten av sänkta ingångslöner.
Vi Moderaterna sätter fokus på en stark välfärd.
Vi Liberalerna skapar tillfällen för smarta investeringar.
Vi Liberalerna vill se tillväxten av fler integrationsåtgärder.
Vi Liberalerna arbetar för en valfrihetsrevolution.
```

## Vidareutveckling med AI

För att visa min kompetens inom AI-integration vidareutvecklade jag skriptet så att det, istället för statisk JSON, anropar en språkmodell via **Microsoft Azure AI Foundry**. Jag ersatte funktionen `flom_load` med `flom_ai_load`, som skickar en systemprompt till modellen och ber den improvisera en satirisk politisk floskel i realtid, i stället för att slumpa fram ord ur en lista.

**Kompetenser jag demonstrerade i det här arbetet:**

- Integration mot Microsoft Azure AI Foundry via `azure-ai-inference`-SDK:et (`ChatCompletionsClient`, `SystemMessage`, `UserMessage`)
- Prompt engineering – en systemprompt som styr ton, språk och tydliga säkerhetsgränser för innehållet
- Autentisering och konfiguration via miljövariabler och `AzureKeyCredential`
- CLI-design med `argparse` för en konfigurerbar körning
- Visualisering – syntaxfärgad kodrendering samt en interaktiv terminalsimulering byggd i React/JSX för att demonstrera hur en skarp körning skulle se ut

````python
# flom.py med AI, simulering av Stefan Blecko 2026-07-11
import os
import argparse
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential


SYSTEM_PROMPT = (
    "Du är en cynisk talskrivare åt svenska politiker. Hitta på EN kort, "
    "absurd och lätt ondskefull politisk floskel på svenska, i stil med "
    "'Vi <parti> vill se <tillväxt> av <något>.' Var satirisk och överdriven, "
    "aldrig hatisk eller kränkande mot verkliga personer. Svara enbart med "
    "meningen, ingen förklaring."
)


def flom_ai_load(client, model="gpt-4o-mini"):
    """Genererar en politisk floskel med hjälp av Microsoft Foundry."""
    response = client.complete(
        model=model,
        temperature=1.0,
        max_tokens=60,
        messages=[
            SystemMessage(content=SYSTEM_PROMPT),
            UserMessage(content="Generera en ny floskel."),
        ],
    )
    return response.choices[0].message.content.strip()


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--antal_floskler", default=1, type=int,
                         help="Antal floskler som ska genereras")
    args = parser.parse_args()

    endpoint = os.environ["AZURE_AI_FOUNDRY_ENDPOINT"]
    api_key = os.environ["AZURE_AI_FOUNDRY_KEY"]
    client = ChatCompletionsClient(endpoint=endpoint,
                                    credential=AzureKeyCredential(api_key))

    for _ in range(args.antal_floskler):
        print(flom_ai_load(client))


if __name__ == "__main__":
    main()
````

**Simulerad körning** — meningarna nedan är förskrivna exempel som visar vad Microsoft Foundry skulle kunna improvisera i en riktig körning (`python flom.py --antal_floskler 5`):

```text

Vi Miljöpartiet räddar klimatet, ett pressmeddelande i taget.
Vi Socialdemokraterna vill se en storslagen utredning om varför ingenting förändras.
Vi Liberalerna älskar frihet, så länge den får plats i rapporten.
Vi Kristdemokraterna välsignar varje beslut med ett pressmeddelande.
```
