### Kapitel 5. Filer, sökvägar, CSV och JSON


JSON är ett vanligt format för konfiguration och strukturerad data. ScriptLab innehåller JSON-ordlistor för AZ-900, DP-900 och MS-102. Exemplet nedan läser in AZ-900-ordlistan om filen finns i den aktuella mappen. Det hanterar medvetet fallet att filen saknas.

### Kapitel 6. Testning, fel och felsökning


Ett undantag (exception) rapporterar ett problem, och validering kan göra orsaken tydlig. Assertioner (assertions) är användbara för snabba kontroller i en handledning. Undvik att tyst undertrycka fel när du behöver förstå varför en åtgärd misslyckades.

````python
def safe_average(values: list[float]) -> float:
    """Returnerar medelvärdet av en icke-tom lista med tal."""
    if not values:
        raise ValueError("Ange minst ett värde.")
    return sum(values) / len(values)

assert safe_average([2, 4]) == 3
safe_average([2, 4, 6])
````

**Prova själv:** Anropa `safe_average([])` och läs felmeddelandet. Lägg till ett test för ett enda värde.

**Det här lärde du dig:** Validera dina antaganden och använd fel för att göra misslyckanden begripliga.

### Kapitel 7. Grunderna i PowerShell och objektpipelinen


PowerShell-kommandon (cmdlets) har ofta namn av typen `Verb-Noun` och skickar objekt – inte bara text – genom pipelinen. `Where-Object` filtrerar objekt och `Select-Object` väljer egenskaper. **PowerShell-exempel** *(kör i PowerShell 7, inte i en Python-cell)*

```powershell
Get-Process |
    Where-Object CPU -GT 10 |
    Select-Object ProcessName, CPU |
    Sort-Object CPU -Descending
```

### Kapitel 8. PowerShell-funktioner och systeminventering


Repots `gn.ps1` visar funktioner och hur man returnerar ett `PSCustomObject` som byggs upp av CIM-data. Börja med en lågpåverkande, skrivskyddad fråga, till exempel uppgifterna om operativsystemet nedan. Dela inte inventeringar som innehåller serienummer, IP-adresser, värdnamn eller andra identifierande uppgifter. Vissa befintliga skript begär administratörsbehörighet eller åtkomst till känslig enhetsinformation; läs och förstå ett skript innan du kör det.

**PowerShell-exempel 1:** en skrivskyddad fråga

```powershell
Get-CimInstance -ClassName Win32_OperatingSystem |
    Select-Object Caption, Version, LastBootUpTime
```

En funktion kan paketera frågan för återanvändning och returnera strukturerad utdata.

**PowerShell-exempel 2:** samma fråga som en funktion

```powershell
function Get-OsSummary {
    Get-CimInstance -ClassName Win32_OperatingSystem |
        Select-Object Caption, Version, LastBootUpTime
}
Get-OsSummary
```

**Prova själv:** Lägg till en parameter i funktionen som styr vilken dator som efterfrågas, och fundera sedan på vilka behörigheter och vilken nätverkskonfiguration det kräver.

### Kapitel 13. Slutprojekt: en liten molninventeringsrapport


Bygg en rapport med fiktiva data först. Om du har en godkänd testtenant och prenumeration kan du ersätta de fiktiva data med godkända, skrivskyddade frågor. Håll rapporten rensad och ta inte med personliga identifierare eller tenantidentifierare.

**Mål**
1. Läs in en JSON-ordlista med Python.
2. Skapa en liten Python-funktion som validerar och sammanfattar en lista med fiktiva resurser.
3. Samla eventuellt in skrivskyddade resultat från Azure eller Microsoft 365 med PowerShell i en separat terminal.
4. Spara bara fiktiva eller rensade data i CSV.
5. Dokumentera behörigheter, beslut om datasäkerhet och hur arbetet kan reproduceras.

**Exempel med fiktiva data**

Följande exempel är helt fiktivt och har inga externa beroenden. Det använder `project_dir` från kapitel 1.

````python
import csv

mock_resources = [
    {"name": "demo-storage-01", "kind": "Storage", "region": "exempelregion"},
    {"name": "demo-app-01", "kind": "App", "region": "exempelregion"},
]

def summarize_resources(resources: list[dict[str, str]]) -> dict[str, int]:
    """Räknar fiktiva resurser per typ."""
    counts: dict[str, int] = {}
    for resource in resources:
        kind = resource["kind"]
        counts[kind] = counts.get(kind, 0) + 1
    return counts

summary = summarize_resources(mock_resources)
print(summary)

report_path = project_dir / "mock-resource-report.csv"
with report_path.open("w", newline="", encoding="utf-8") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=["name", "kind", "region"])
    writer.writeheader()
    writer.writerows(mock_resources)

print(f"Skrev en fiktiv exempelrapport till {report_path}")
````
